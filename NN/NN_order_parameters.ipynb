```python
### averaged profile (all chains, Lo and Ld together) for each system
def all_chain_profile(lower_ops, upper_ops):
    return np.concatenate([lower_ops.reshape(-1, 16), upper_ops.reshape(-1, 16)]).mean(axis=0)

sn1_av_65 = all_chain_profile(dopc_65_sn1_ops_lower_5pt, dopc_65_sn1_ops_upper_5pt)
sn2_av_65 = all_chain_profile(dopc_65_sn2_ops_lower_5pt, dopc_65_sn2_ops_upper_5pt)
sn1_av_75 = all_chain_profile(dopc_75_sn1_ops_lower_5pt, dopc_75_sn1_ops_upper_5pt)
sn2_av_75 = all_chain_profile(dopc_75_sn2_ops_lower_5pt, dopc_75_sn2_ops_upper_5pt)
sn1_av_90 = all_chain_profile(dopc_90_sn1_ops_lower_5pt, dopc_90_sn1_ops_upper_5pt)
sn2_av_90 = all_chain_profile(dopc_90_sn2_ops_lower_5pt, dopc_90_sn2_ops_upper_5pt)
```

### Uncertainty: χ_C = 0.25 replicas

```python
### chi_C = 0.25 replicas: the same calculation as for dopc_75 above (300 DOPC per leaflet)
def lo_ld_profiles(sys_name, n_point=5):
    """sn1 / sn2 Lo and Ld mean profiles of one system, calculated as in the cells above."""
    pos_lower, pos_upper = clf_lo_ld_by_chain(sys_name , sys_type , "raw" , "nn" , "sn1" , 3 , start_frame , stop_frame).bilayer_all_dopc_data()
    out = {}
    for chain in ("sn1", "sn2"):
        labels_lower, labels_upper = np.load(os.path.join(data_array_dir , f"{sys_name}_{chain}_ft_last_{n_point}pt_prediction_labels.npy") , allow_pickle=True)
        ops = []
        for pos in (pos_lower, pos_upper):
            n_lipids = pos.shape[1]
            raw = np.empty([stop_frame, n_lipids, 16])
            for i in range(stop_frame):
                for j in range(n_lipids):
                    raw[i, j] = DOPC_op_H_av(pos[i, j], chain)
            ops.append(convolve_ops(raw , n_point))
        lo_lower, ld_lower = split_ops_by_label(ops[0], labels_lower)
        lo_upper, ld_upper = split_ops_by_label(ops[1], labels_upper)
        lo_mean, lo_std, ld_mean, ld_std = mean_op_profile(lo_upper, lo_lower, ld_upper, ld_lower)
        out[chain] = (lo_mean, ld_mean)
    return out

rep1 = lo_ld_profiles("dopc_75_rep1")
rep2 = lo_ld_profiles("dopc_75_rep2")

### replica SD (ddof = 1) per carbon over dopc_75, dopc_75_rep1, dopc_75_rep2
replica_sd = {}
for chain, lo_75, ld_75 in (("sn1", sn1_lo_mean_75, sn1_ld_mean_75), ("sn2", sn2_lo_mean_75, sn2_ld_mean_75)):
    replica_sd[f"{chain}_lo"] = np.std([lo_75, rep1[chain][0], rep2[chain][0]], axis=0, ddof=1)
    replica_sd[f"{chain}_ld"] = np.std([ld_75, rep1[chain][1], rep2[chain][1]], axis=0, ddof=1)
for k, v in replica_sd.items():
    print(f"{k}: replica SD {v.min():.5f} - {v.max():.5f}")
```

### Figure (paper style)

```python
# paper figure style: bold text, ticks on all four sides pointing in, minor ticks
PAPER_STYLE = {"font.weight": "bold", "axes.labelweight": "bold", "axes.titleweight": "bold", "axes.linewidth": 2.5,
               "mathtext.default": "bf", "xtick.top": True, "ytick.right": True, "xtick.direction": "in",
               "ytick.direction": "in", "xtick.major.size": 10, "ytick.major.size": 10, "xtick.major.width": 2.5,
               "ytick.major.width": 2.5, "xtick.minor.size": 5, "ytick.minor.size": 5, "xtick.minor.width": 1.8,
               "ytick.minor.width": 1.8, "xtick.major.pad": 8, "ytick.major.pad": 8}
SN = {"sn1": r"$\mathbfit{sn}$-1", "sn2": r"$\mathbfit{sn}$-2"}
LO, LD = r"$\mathbf{L_o}$", r"$\mathbf{L_d}$"


def style_axes(ax, size=20):
    ax.xaxis.set_minor_locator(AutoMinorLocator(2))
    ax.yaxis.set_minor_locator(AutoMinorLocator(2))
    ax.tick_params(axis="both", which="both", labelsize=size, top=True, right=True, direction="in")
    for lab in ax.get_xticklabels() + ax.get_yticklabels():
        lab.set_fontweight("bold")


def legend_without_error_bars(ax, **kwargs):
    """Legend that shows only the line and marker of each curve, not its error bar."""
    handles, labels = ax.get_legend_handles_labels()
    handles = [h[0] if isinstance(h, ErrorbarContainer) else h for h in handles]
    return ax.legend(handles, labels, **kwargs)


ERR = dict(capsize=8, elinewidth=2.5, capthick=2.5, markersize=10)
out_dir = "../heavy_data_NO_GIT/results/op_profiles"
os.makedirs(out_dir, exist_ok=True)

OP = {"dopc_90": {"sn1": (sn1_lo_mean_90, sn1_ld_mean_90, sn1_av_90), "sn2": (sn2_lo_mean_90, sn2_ld_mean_90, sn2_av_90)},
      "dopc_75": {"sn1": (sn1_lo_mean_75, sn1_ld_mean_75, sn1_av_75), "sn2": (sn2_lo_mean_75, sn2_ld_mean_75, sn2_av_75)},
      "dopc_65": {"sn1": (sn1_lo_mean_65, sn1_ld_mean_65, sn1_av_65), "sn2": (sn2_lo_mean_65, sn2_ld_mean_65, sn2_av_65)}}
CHOL = {"dopc_90": 0.10, "dopc_75": 0.25, "dopc_65": 0.35}
carbons = np.arange(2, 18)                                   # OP index 1..16 = acyl carbons C2..C17
tick_carbons = carbons[carbons % 2 == 0]
panel_labels = "cdefgh"

with plt.rc_context(PAPER_STYLE):
    fig, axes = plt.subplots(3, 2, figsize=(30, 27), sharey=True, squeeze=False)
    for row, sys_name in enumerate(OP):
        chi = rf"$\mathbf{{\chi_C = {CHOL[sys_name]:g}}}$"
        for col, chain in enumerate(("sn1", "sn2")):
            ax = axes[row, col]
            lo, ld, av = OP[sys_name][chain]
            # error bars only at chi_C = 0.25 (where the replicas are)
            at_025 = CHOL[sys_name] == 0.25
            ax.errorbar(carbons, lo, yerr=replica_sd[f"{chain}_lo"] if at_025 else None, fmt="o-", color="red",
                        linewidth=4, label=LO, **ERR)
            ax.errorbar(carbons, ld, yerr=replica_sd[f"{chain}_ld"] if at_025 else None, fmt="o-", color="green",
                        linewidth=4, label=LD, **ERR)
            ax.errorbar(carbons, av, fmt="o", linestyle=":", color="black", linewidth=4, label="Averaged", **ERR)
            ax.set_title(f"{SN[chain]} | {chi}", fontsize=35)
            ax.text(0.5, 0.98, panel_labels[row * 2 + col], transform=ax.transAxes, fontsize=48,
                    fontweight="bold", ha="center", va="top", color="black")
            ax.set_ylim(0, 0.38)
            style_axes(ax, size=35)
            ax.set_xticks(tick_carbons, [f"C{c}" for c in tick_carbons], fontsize=35, fontweight="bold")
            ax.set_xticks(carbons[carbons % 2 == 1], minor=True)
            # legend: error bars shown only where the curves have them (Lo, Ld at chi_C = 0.25)
            if at_025:
                handles, labels = ax.get_legend_handles_labels()
                handles = [h if lab != "Averaged" else h[0] for h, lab in zip(handles, labels)]
                ax.legend(handles, labels, fontsize=35, frameon=False, loc="upper right")
            else:
                legend_without_error_bars(ax, fontsize=35, frameon=False, loc="upper right")
            if col == 0:
                ax.set_ylabel(r"$\mathbf{S_{CH}}$", fontsize=45)
    for col in range(2):
        axes[-1, col].set_xlabel("Carbon Atom", fontsize=35)
    fig.tight_layout()
    fig.savefig(os.path.join(out_dir, "nn_5pt_op_profiles.png"), dpi=150)
    plt.show()
```
