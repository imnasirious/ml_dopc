### Lo/Ld fractions

```python
chol_fracs = np.array([0.0625 , 0.10, 0.15, 0.20, 0.25, 0.30, 0.35])
chol_fracs_all = np.array([0 , 0.0625, 0.10, 0.15, 0.20, 0.25, 0.30, 0.35 , 0.4])
systems_17pt_sn1 = [dopc_94_17pt_sn1, dopc_90_17pt_sn1, dopc_85_17pt_sn1, dopc_80_17pt_sn1, dopc_75_17pt_sn1, dopc_70_17pt_sn1, dopc_65_17pt_sn1]
systems_17pt_sn2 = [dopc_94_17pt_sn2, dopc_90_17pt_sn2, dopc_85_17pt_sn2, dopc_80_17pt_sn2, dopc_75_17pt_sn2, dopc_70_17pt_sn2, dopc_65_17pt_sn2]

no_frames = dopc_94_17pt_sn1.prediction_labels[0].shape[0]

lower_means_17pt_sn1 = []
upper_means_17pt_sn1 = []

lower_means_17pt_sn2 = []
upper_means_17pt_sn2 = []

for sys in systems_17pt_sn1:
    sys.frames_out = no_frames                       # labels were loaded from disk, not made by process_all_input()
    lower_sn1, upper_sn1 = sys.lo_ld_proportions()   # (lower_array, upper_array)
    lower_means_17pt_sn1.append(np.mean(lower_sn1))
    upper_means_17pt_sn1.append(np.mean(upper_sn1))

for sys in systems_17pt_sn2:
    sys.frames_out = no_frames
    lower_sn2, upper_sn2 = sys.lo_ld_proportions()   # (lower_array, upper_array)
    lower_means_17pt_sn2.append(np.mean(lower_sn2))
    upper_means_17pt_sn2.append(np.mean(upper_sn2))

lower_upper_means_17pt_sn1_av = 0.5 * (np.array(lower_means_17pt_sn1) + np.array(upper_means_17pt_sn1))
lower_upper_means_17pt_sn2_av = 0.5 * (np.array(lower_means_17pt_sn2) + np.array(upper_means_17pt_sn2))
```

### Endpoints (upper leaflet only)

```python
### endpoints: dopc_pure (chi_C = 0) and dopc_60 (chi_C = 0.40)
### only the upper leaflet is used, because the lower leaflet of these two systems was the training leaflet

endpoints_17pt_sn1 = [dopc_pure_17pt_sn1, dopc_60_17pt_sn1]
endpoints_17pt_sn2 = [dopc_pure_17pt_sn2, dopc_60_17pt_sn2]

endpoint_means_17pt_sn1 = []
endpoint_means_17pt_sn2 = []

for sys in endpoints_17pt_sn1:
    sys.frames_out = no_frames
    lower_sn1, upper_sn1 = sys.lo_ld_proportions()
    endpoint_means_17pt_sn1.append(np.mean(upper_sn1))

for sys in endpoints_17pt_sn2:
    sys.frames_out = no_frames
    lower_sn2, upper_sn2 = sys.lo_ld_proportions()
    endpoint_means_17pt_sn2.append(np.mean(upper_sn2))

### all points from chi_C = 0 to 0.40 (same order as chol_fracs_all)
means_all_17pt_sn1 = np.concatenate([[endpoint_means_17pt_sn1[0]], lower_upper_means_17pt_sn1_av, [endpoint_means_17pt_sn1[1]]])
means_all_17pt_sn2 = np.concatenate([[endpoint_means_17pt_sn2[0]], lower_upper_means_17pt_sn2_av, [endpoint_means_17pt_sn2[1]]])
```

### Uncertainty from the χ_C = 0.25 replicas

```python
### uncertainty: SD (ddof = 1) over the three chi_C = 0.25 runs (dopc_75, dopc_75_rep1, dopc_75_rep2);
### shown as an error bar at chi_C = 0.25 only

replicas_17pt_sn1 = [dopc_75_17pt_sn1, dopc_75_rep1_17pt_sn1, dopc_75_rep2_17pt_sn1]
replicas_17pt_sn2 = [dopc_75_17pt_sn2, dopc_75_rep1_17pt_sn2, dopc_75_rep2_17pt_sn2]

replica_means_17pt_sn1 = []
replica_means_17pt_sn2 = []

for sys in replicas_17pt_sn1:
    sys.frames_out = no_frames
    lower_sn1, upper_sn1 = sys.lo_ld_proportions()
    replica_means_17pt_sn1.append(0.5 * (np.mean(lower_sn1) + np.mean(upper_sn1)))

for sys in replicas_17pt_sn2:
    sys.frames_out = no_frames
    lower_sn2, upper_sn2 = sys.lo_ld_proportions()
    replica_means_17pt_sn2.append(0.5 * (np.mean(lower_sn2) + np.mean(upper_sn2)))

replica_sd_17pt_sn1 = np.std(replica_means_17pt_sn1, ddof=1)
replica_sd_17pt_sn2 = np.std(replica_means_17pt_sn2, ddof=1)
```

### XGB 17pt

```python
print(f"replica SD at chi_C = 0.25: sn1 {replica_sd_17pt_sn1:.4f}, sn2 {replica_sd_17pt_sn2:.4f}\n")
print(f"{'Chol':>6} | {'sn1 Lo frac':>11} | {'sn2 Lo frac':>11}")
print("-" * 36)
for chol, f1, f2 in zip(chol_fracs_all, means_all_17pt_sn1, means_all_17pt_sn2):
    print(f"{chol:>6.4f} | {f1:>11.4f} | {f2:>11.4f}")
```

```python
# paper figure style: bold text, ticks on all four sides pointing in, minor ticks
PAPER_STYLE = {"font.weight": "bold", "axes.labelweight": "bold", "axes.titleweight": "bold", "axes.linewidth": 2.5,
               "mathtext.default": "bf", "xtick.top": True, "ytick.right": True, "xtick.direction": "in",
               "ytick.direction": "in", "xtick.major.size": 10, "ytick.major.size": 10, "xtick.major.width": 2.5,
               "ytick.major.width": 2.5, "xtick.minor.size": 5, "ytick.minor.size": 5, "xtick.minor.width": 1.8,
               "ytick.minor.width": 1.8, "xtick.major.pad": 8, "ytick.major.pad": 8}
SN = {"sn1": r"$\mathbfit{sn}$-1", "sn2": r"$\mathbfit{sn}$-2"}
LO, LD = r"$\mathbf{L_o}$", r"$\mathbf{L_d}$"


def style_axes(ax, size=20):
    ax.xaxis.set_minor_locator(AutoMinorLocator(2))
    ax.yaxis.set_minor_locator(AutoMinorLocator(2))
    ax.tick_params(axis="both", which="both", labelsize=size, top=True, right=True, direction="in")
    for lab in ax.get_xticklabels() + ax.get_yticklabels():
        lab.set_fontweight("bold")


def legend_without_error_bars(ax, **kwargs):
    """Legend that shows only the line and marker of each curve, not its error bar."""
    handles, labels = ax.get_legend_handles_labels()
    handles = [h[0] if isinstance(h, ErrorbarContainer) else h for h in handles]
    return ax.legend(handles, labels, **kwargs)


ERR = dict(capsize=8, elinewidth=2.5, capthick=2.5, markersize=12)
out_dir = "../heavy_data_NO_GIT/results/lo_ld_fractions"
os.makedirs(out_dir, exist_ok=True)

with plt.rc_context(PAPER_STYLE):
    fig, ax = plt.subplots(figsize=(12, 10))
    for means, sd, color, chain in ((means_all_17pt_sn1, replica_sd_17pt_sn1, "green", "sn1"),
                                    (means_all_17pt_sn2, replica_sd_17pt_sn2, "red", "sn2")):
        # error bar only at chi_C = 0.25 (where the replicas are); NaN = no error bar
        ax.errorbar(chol_fracs_all, means, yerr=np.where(np.isclose(chol_fracs_all, 0.25), sd, np.nan),
                    fmt="o", color=color, label=SN[chain], **ERR)
    ax.set_xlim(-0.02, 0.42)
    ax.set_ylim(0, 1.0)
    ax.set_xlabel("CHOL fraction", fontsize=28)
    ax.set_ylabel(rf"Fraction of {LO} chains", fontsize=28)
    ax.text(0.03, 0.97, "XGB", transform=ax.transAxes, fontsize=50, fontweight="bold", ha="left", va="top")
    style_axes(ax, size=32)
    legend_without_error_bars(ax, loc="lower right", fontsize=45, frameon=False)
    fig.tight_layout()
    fig.savefig(os.path.join(out_dir, "xgb_17pt_lo_fractions.png"), dpi=200)
    plt.show()
```
