```python
def avg_bilayer_means(lower_sn1, lower_sn2, upper_sn1, upper_sn2) -> dict:

    lo_l = np.sum(lower_sn1 & lower_sn2, axis=1)
    ld_l = np.sum((~lower_sn1) & (~lower_sn2), axis=1)
    mx_l = np.sum(lower_sn1 ^ lower_sn2, axis=1)
    lo_u = np.sum(upper_sn1 & upper_sn2, axis=1)
    ld_u = np.sum((~upper_sn1) & (~upper_sn2), axis=1)
    mx_u = np.sum(upper_sn1 ^ upper_sn2, axis=1)
    lo_t = lo_l + lo_u
    ld_t = ld_l + ld_u
    mx_t = mx_l + mx_u

    # ---- Chain state counts per frame (sn1+sn2, both leaflets)
    lo_chains_t = (
        np.sum(lower_sn1, axis=1) +
        np.sum(lower_sn2, axis=1) +
        np.sum(upper_sn1, axis=1) +
        np.sum(upper_sn2, axis=1)
    )
    total_lipids = lower_sn1.shape[1] + upper_sn1.shape[1]
    total_chains = 2 * total_lipids
    ld_chains_t = total_chains - lo_chains_t

    # ---- Means (counts)
    lo_mean = float(lo_t.mean())
    ld_mean = float(ld_t.mean())
    mx_mean = float(mx_t.mean())
    lo_ch_mean = float(lo_chains_t.mean())
    ld_ch_mean = float(ld_chains_t.mean())

    # ---- Standard deviations (counts)
    lo_std = float(lo_t.std())
    ld_std = float(ld_t.std())
    mx_std = float(mx_t.std())
    lo_ch_std = float(lo_chains_t.std())
    ld_ch_std = float(ld_chains_t.std())

    # ---- Percentages (floats)
    lo_pct     = 100.0 * lo_mean    / total_lipids
    ld_pct     = 100.0 * ld_mean    / total_lipids
    mx_pct     = 100.0 * mx_mean    / total_lipids
    lo_ch_pct  = 100.0 * lo_ch_mean / total_chains
    ld_ch_pct  = 100.0 * ld_ch_mean / total_chains

    # ---- Percentage standard deviations (floats)
    lo_pct_std    = 100.0 * lo_std    / total_lipids
    ld_pct_std    = 100.0 * ld_std    / total_lipids
    mx_pct_std    = 100.0 * mx_std    / total_lipids
    lo_ch_pct_std = 100.0 * lo_ch_std / total_chains
    ld_ch_pct_std = 100.0 * ld_ch_std / total_chains

    #---------------------------------------------- NEW: Mixed sub-categories (total bilayer) ---------------------------------------------------#
    # Mixed_sn1Lo : sn1 is ordered, sn2 is disordered
    # Mixed_sn2Lo : sn2 is ordered, sn1 is disordered

    mx_sn1lo_l = np.sum(lower_sn1 & ~lower_sn2, axis=1)   # lower leaflet: sn1=Lo, sn2=Ld
    mx_sn2lo_l = np.sum(~lower_sn1 & lower_sn2, axis=1)   # lower leaflet: sn1=Ld, sn2=Lo

    mx_sn1lo_u = np.sum(upper_sn1 & ~upper_sn2, axis=1)   # upper leaflet: sn1=Lo, sn2=Ld
    mx_sn2lo_u = np.sum(~upper_sn1 & upper_sn2, axis=1)   # upper leaflet: sn1=Ld, sn2=Lo

    mx_sn1lo_t = mx_sn1lo_l + mx_sn1lo_u   # total bilayer
    mx_sn2lo_t = mx_sn2lo_l + mx_sn2lo_u   # total bilayer

    mx_sn1lo_mean = float(mx_sn1lo_t.mean())
    mx_sn2lo_mean = float(mx_sn2lo_t.mean())

    mx_sn1lo_std = float(mx_sn1lo_t.std())
    mx_sn2lo_std = float(mx_sn2lo_t.std())

    mx_sn1lo_pct = 100.0 * mx_sn1lo_mean / mx_mean if mx_mean > 0 else 0.0
    mx_sn2lo_pct = 100.0 * mx_sn2lo_mean / mx_mean if mx_mean > 0 else 0.0

    mx_sn1lo_pct_std = 100.0 * mx_sn1lo_std / mx_mean if mx_mean > 0 else 0.0
    mx_sn2lo_pct_std = 100.0 * mx_sn2lo_std / mx_mean if mx_mean > 0 else 0.0


    return {
        "Both_Chains_LO%":     lo_pct,
        "Both_Chains_LO%_std": lo_pct_std,
        "Both_Chains_LD%":     ld_pct,
        "Both_Chains_LD%_std": ld_pct_std,
        "Mixed_Chains%":       mx_pct,
        "Mixed_Chains%_std":   mx_pct_std,
        "LO_Chains%":          lo_ch_pct,
        "LO_Chains%_std":      lo_ch_pct_std,
        "LD_Chains%":          ld_ch_pct,
        "LD_Chains%_std":      ld_ch_pct_std,
        ###----------New----------------------#
        "Mixed_sn1Lo": mx_sn1lo_mean,
        "Mixed_sn2Lo": mx_sn2lo_mean,
        "Mixed_sn1Lo_pct": mx_sn1lo_pct,
        "Mixed_sn2Lo_pct": mx_sn2lo_pct,
        "Mixed_sn1Lo%_std":mx_sn1lo_pct_std,
        "Mixed_sn2Lo%_std":mx_sn2lo_pct_std,
        
    }

    return {
        "Both_Chains_LO%":     lo_pct,
        "Both_Chains_LO%_std": lo_pct_std,
        "Both_Chains_LD%":     ld_pct,
        "Both_Chains_LD%_std": ld_pct_std,
        "Mixed_Chains%":       mx_pct,
        "Mixed_Chains%_std":   mx_pct_std,
        "LO_Chains%":          lo_ch_pct,
        "LO_Chains%_std":      lo_ch_pct_std,
        "LD_Chains%":          ld_ch_pct,
        "LD_Chains%_std":      ld_ch_pct_std,
    }
```

```python
def three_state_entropy(both_lo_pct, both_ld_pct, mixed_pct):

    entropy_list = []
    
    for lo, ld, mx in zip(both_lo_pct, both_ld_pct, mixed_pct):
        
        p_lo = lo / 100.0
        p_ld = ld / 100.0
        p_mx = mx / 100.0
        
        # guard against log(0)
        terms = []
        for p in [p_lo, p_ld, p_mx]:
            if p > 0:
                terms.append(-p * np.log(p))
            else:
                terms.append(0.0)
        
        entropy_list.append(sum(terms))
    
    return np.array(entropy_list)

```
