## Velocity autocorrelation functions

```python
from scipy.signal import savgol_filter
```

```python
def velocity(com_array , ts_in_ps):

    original_time  = len(com_array)
    no_lipids = com_array.shape[1]

    ts_in_ns = ts_in_ps/1000
    

    velocities = np.empty([original_time-1 , no_lipids , 3])

    for ts in range(1 , original_time):

        position_current , position_next = com_array[ts-1] , com_array[ts]
        vel = (position_next - position_current) / ts_in_ns
        velocities[ts-1] = vel

    return velocities


def velocity_autocorrelation(velocity_array, lags):
    frame_0 = velocity_array[0]            # (n_atoms, 3)
    no_atoms = velocity_array.shape[1]

    vacf = []
    for t in lags:
        frame_t = velocity_array[t]
        dot_sum = np.sum(np.sum(frame_0 * frame_t, axis=1))
        vacf.append(dot_sum / no_atoms)

    vacf = np.array(vacf)
    return vacf / vacf[0]   # normalize by C(0)


def tau_integral(lags, vacf, cutoff='first_zero'):

    lags = np.asarray(lags)
    vacf = np.asarray(vacf)

    if cutoff == 'first_zero':

        zero_idx = np.where(vacf <= 0)[0]
        end_idx = zero_idx[0] if len(zero_idx) else len(vacf) - 1
    elif isinstance(cutoff, (int, float)):
        end_idx = np.searchsorted(lags, cutoff)
        end_idx = min(end_idx, len(lags) - 1)
    else:
        end_idx = len(vacf) - 1

    tau = np.trapezoid(vacf[:end_idx+1], lags[:end_idx+1])
    return tau
        
```

```python
start_frame , stop_frame = 0 , 5000

sn1_indices = [41, 42 , 43]
sn1_indices.extend(i for i in range(91 ,138))

sn2_indices = [32, 33 , 34]
sn2_indices.extend(i for i in range(44 ,91))

lags = [i for i in range(101)]
lag_times = [10*i for i in lags]
```

```python
dopc_pure_raw = Bilayer("dopc_no_chol_raw" , "sym" , start_frame , stop_frame)
dopc_pure_raw.create_universe()
dopc_60_raw = Bilayer("dopc_60_raw" , "sym" , start_frame , stop_frame)
dopc_60_raw.create_universe()

dopc_pure_raw.save_residue_positions("DOPC")   ####RUN ONCE
dopc_60_raw.save_residue_positions("DOPC")
dopc_60_raw.save_residue_positions("CHL1")
```

```python
pure_dopc_raw.cleave_leaflets()
dopc_60_raw.cleave_leaflets()
```

```python
dopc_pure_lower_leaflet = dopc_pure_raw.lower_leaflet
dopc_pure_upper_leaflet = dopc_pure_raw.upper_leaflet

dopc_pure_lower_sn1 = dopc_pure_lower_leaflet[: , : , sn1_indices , :]
dopc_pure_upper_sn1 = dopc_pure_upper_leaflet[: , : , sn1_indices , :]

dopc_pure_lower_sn2 = dopc_pure_lower_leaflet[: , : , sn2_indices , :]
dopc_pure_upper_sn2 = dopc_pure_upper_leaflet[: , : , sn2_indices , :]

no_lipids = dopc_pure_lower_leaflet.shape[1]

com_sn1_all_frames_lower = np.empty([stop_frame , no_lipids , 3])
com_sn1_all_frames_upper = np.empty([stop_frame , no_lipids , 3])
com_sn2_all_frames_lower = np.empty([stop_frame , no_lipids , 3])
com_sn2_all_frames_upper = np.empty([stop_frame , no_lipids , 3])

for ts in range(stop_frame):

    one_frame_com_lower_sn1 = dopc_pure_lower_sn1[ts] 
    one_frame_com_upper_sn1 = dopc_pure_upper_sn1[ts]
    one_frame_com_lower_sn2 = dopc_pure_lower_sn2[ts]
    one_frame_com_upper_sn2 = dopc_pure_upper_sn2[ts]

    for lipid in range(no_lipids):
        
        one_frame_com_lower_sn1_lipid = com_one_acyl_chain("sn1" , one_frame_com_lower_sn1[lipid , : , :])
        one_frame_com_upper_sn1_lipid = com_one_acyl_chain("sn1" , one_frame_com_upper_sn1[lipid , : , :])
        one_frame_com_lower_sn2_lipid = com_one_acyl_chain("sn2" , one_frame_com_lower_sn2[lipid, : , :])
        one_frame_com_upper_sn2_lipid = com_one_acyl_chain("sn2" ,one_frame_com_upper_sn2[lipid , : , :])

        com_sn1_all_frames_lower[ts , lipid ] = one_frame_com_lower_sn1_lipid
        com_sn1_all_frames_upper[ts , lipid] = one_frame_com_upper_sn1_lipid
        com_sn2_all_frames_lower[ts , lipid] = one_frame_com_lower_sn2_lipid
        com_sn2_all_frames_upper[ts , lipid] = one_frame_com_upper_sn2_lipid

dopc_pure_com_acyl_lower = (com_sn1_all_frames_lower + com_sn2_all_frames_lower)*0.5
dopc_pure_com_acyl_upper = (com_sn1_all_frames_upper + com_sn2_all_frames_upper)*0.5
```

```python
dopc_60_lower_leaflet = dopc_60_raw.lower_leaflet
dopc_60_upper_leaflet = dopc_60_raw.upper_leaflet

dopc_60_lower_sn1 = dopc_60_lower_leaflet[0][: , : , sn1_indices , :]
dopc_60_upper_sn1 = dopc_60_upper_leaflet[0][: , : , sn1_indices , :]

dopc_60_lower_sn2 = dopc_60_lower_leaflet[0][: , : , sn2_indices , :]
dopc_60_upper_sn2 = dopc_60_upper_leaflet[0][: , : , sn2_indices , :]

no_lipids = dopc_60_lower_leaflet[0].shape[1]

com_sn1_all_frames_lower = np.empty([stop_frame , no_lipids , 3])
com_sn1_all_frames_upper = np.empty([stop_frame , no_lipids , 3])
com_sn2_all_frames_lower = np.empty([stop_frame , no_lipids , 3])
com_sn2_all_frames_upper = np.empty([stop_frame , no_lipids , 3])

for ts in range(stop_frame):

    one_frame_com_lower_sn1 = dopc_60_lower_sn1[ts] 
    one_frame_com_upper_sn1 = dopc_60_upper_sn1[ts]
    one_frame_com_lower_sn2 = dopc_60_lower_sn2[ts]
    one_frame_com_upper_sn2 = dopc_60_upper_sn2[ts]

    for lipid in range(no_lipids):
        
        one_frame_com_lower_sn1_lipid = com_one_acyl_chain("sn1" , one_frame_com_lower_sn1[lipid , : , :])
        one_frame_com_upper_sn1_lipid = com_one_acyl_chain("sn1" , one_frame_com_upper_sn1[lipid , : , :])
        one_frame_com_lower_sn2_lipid = com_one_acyl_chain("sn2" , one_frame_com_lower_sn2[lipid, : , :])
        one_frame_com_upper_sn2_lipid = com_one_acyl_chain("sn2" ,one_frame_com_upper_sn2[lipid , : , :])

        com_sn1_all_frames_lower[ts , lipid ] = one_frame_com_lower_sn1_lipid
        com_sn1_all_frames_upper[ts , lipid] = one_frame_com_upper_sn1_lipid
        com_sn2_all_frames_lower[ts , lipid] = one_frame_com_lower_sn2_lipid
        com_sn2_all_frames_upper[ts , lipid] = one_frame_com_upper_sn2_lipid

dopc_60_com_acyl_lower = (com_sn1_all_frames_lower + com_sn2_all_frames_lower)*0.5
dopc_60_com_acyl_upper = (com_sn1_all_frames_upper + com_sn2_all_frames_upper)*0.5
```

```python
vel_dopc_pure_lower = velocity(dopc_pure_com_acyl_lower , 10)
vel_dopc_pure_upper = velocity(dopc_pure_com_acyl_upper , 10)

vel_dopc_pure_av = (vel_dopc_pure_lower + vel_dopc_pure_upper)*0.5


cor_dopc_pure = velocity_autocorrelation(vel_dopc_pure_av , lags)

vel_dopc_60_lower = velocity(dopc_60_com_acyl_lower , 10)
vel_dopc_60_upper = velocity(dopc_60_com_acyl_upper , 10)

vel_dopc_60_av = (vel_dopc_60_lower + vel_dopc_60_upper)*0.5


cor_dopc_60 = velocity_autocorrelation(vel_dopc_60_av , lags)

cor_dopc_pure_smooth = savgol_filter(cor_dopc_pure, window_length=40, polyorder=13)
cor_dopc_60_smooth = savgol_filter(cor_dopc_60, window_length=40, polyorder=13)


plt.figure(figsize=[10,8])
plt.plot(lag_times , cor_dopc_pure_smooth)
plt.plot(lag_times, cor_dopc_60_smooth)
plt.legend(["Pure DOPC" , "DOPC 60%"])

plt.ylabel("Velocity Autocorrelation")
plt.xlabel("Lag time / ps")
plt.ylim(-0.5 , 1.0)
plt.xlim(0 , 100)

plt.show()
```

```python
tau_integral(lag_times , cor_dopc_pure_smooth , cutoff= "first_zero")
```
