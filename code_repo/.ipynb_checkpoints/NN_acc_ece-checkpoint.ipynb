```python
prob_threshold = 0.5
```

```python
def ece(y_true, y_prob, n_bins):
    """
    Calculates the Expected Calibration Error (ECE) for a binary classifier.
    """
    # Get the fraction of positives (true probabilities) and mean predicted values for each bin
    prob_true, prob_pred = calibration_curve(y_true, y_prob, n_bins=n_bins, strategy='uniform')
    
    # Calculate the proportion of samples in each bin
    bin_total = np.histogram(y_prob, bins=n_bins, range=(0., 1.))[0]
    bin_proportions = bin_total / len(y_true)
    
    # Calculate ECE as the weighted average of absolute differences
    ece = np.sum(np.abs(prob_true - prob_pred) * bin_proportions)
    
    return ece



def min_max_normalize(arr):
    arr = np.asarray(arr, dtype=float) # Convert to float array
    return (arr - arr.min()) / (arr.max() - arr.min())

        

        
```

## NEW SELECTION CRITERION:

\begin{equation}
    n^{*} = \underset{n}{\mathrm{argmin}} \left( \frac{|ECE_{\mathrm{sn1}}(n) - ECE_{\mathrm{sn2}}(n)|}{ACC_{\mathrm{sn1}}(n) + ACC_{\mathrm{sn2}}(n)} \right)
\end{equation}

```python
ece_sn2
```

```python
###implemented here. pick the window corresponding to smallest number.

n_stars = np.abs(np.array(ece_sn1) - np.array(ece_sn2))/(np.array(accuracies_sn1) + np.array(accuracies_sn2))
n_stars
```
