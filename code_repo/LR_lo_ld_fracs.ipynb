### Lo/Ld fractions

```python
chol_fracs = np.array([0.0625 , 0.10, 0.15, 0.20, 0.25, 0.30, 0.35])
chol_fracs_all = np.array([0 , 0.0625, 0.10, 0.15, 0.20, 0.25, 0.30, 0.35 , 0.4])
endpoint_0 = np.array([0 , 0.1])
endpoint_40 = np.array([0.35 , 0.4])
systems_15pt_sn1 = [dopc_94_15pt_sn1, dopc_90_15pt_sn1, dopc_85_15pt_sn1, dopc_80_15pt_sn1, dopc_75_15pt_sn1, dopc_70_15pt_sn1, dopc_65_15pt_sn1]
systems_15pt_sn2 = [dopc_94_15pt_sn2, dopc_90_15pt_sn2, dopc_85_15pt_sn2, dopc_80_15pt_sn2, dopc_75_15pt_sn2, dopc_70_15pt_sn2, dopc_65_15pt_sn2]

no_frames = dopc_94_15pt_sn1.lo_ld_proportions()[0].shape[0]

lower_means_15pt_sn1 = []
upper_means_15pt_sn1 = []
lower_sds_15pt_sn1 = []
upper_sds_15pt_sn1 = []

lower_means_15pt_sn2 = []
upper_means_15pt_sn2 = []
lower_sds_15pt_sn2 = []
upper_sds_15pt_sn2 = []

for sys in systems_15pt_sn1:
    lower_sn1, upper_sn1 = sys.lo_ld_proportions()   # (lower_array, upper_array)
    lower_means_15pt_sn1.append(np.mean(lower_sn1))
    upper_means_15pt_sn1.append(np.mean(upper_sn1))
    lower_sds_15pt_sn1.append((np.std(lower_sn1)))
    upper_sds_15pt_sn1.append(np.std(upper_sn1))

for sys in systems_15pt_sn2:
    lower_sn2, upper_sn2 = sys.lo_ld_proportions()   # (lower_array, upper_array)
    lower_means_15pt_sn2.append(np.mean(lower_sn2))
    upper_means_15pt_sn2.append(np.mean(upper_sn2))
    lower_sds_15pt_sn2.append((np.std(lower_sn2)))
    upper_sds_15pt_sn2.append(np.std(upper_sn2))

lower_upper_means_15pt_sn1_av = 0.5 * (np.array(lower_means_15pt_sn1) + np.array(upper_means_15pt_sn1))
lower_upper_sds_15pt_sn1_av = 0.5 * (np.array(lower_sds_15pt_sn1) + np.array(upper_sds_15pt_sn1))
lower_upper_ses_15pt_sn1_av = [sd / np.sqrt(no_frames) for sd in lower_upper_sds_15pt_sn1_av]

lower_upper_means_15pt_sn2_av = 0.5 * (np.array(lower_means_15pt_sn2) + np.array(upper_means_15pt_sn2))
lower_upper_sds_15pt_sn2_av = 0.5 * (np.array(lower_sds_15pt_sn2) + np.array(upper_sds_15pt_sn2))
lower_upper_ses_15pt_sn2_av = [sd / np.sqrt(no_frames) for sd in lower_upper_sds_15pt_sn2_av]

```

## SIGMOID FITS

```python
### sn1

initial_guesses = [-1 , 0.5 , 0.2 , 1]

popt_sn1, pcov_sn1 = curve_fit(sigmoid, chol_fracs, lower_upper_means_15pt_sn1_av, p0=initial_guesses)

a_opt_sn1, b_opt_sn1, c_opt_sn1 , d_opt_sn1 = popt_sn1
print(f"Optimized parameters: a={a_opt_sn1:.4f}, b={b_opt_sn1:.4f}, c={c_opt_sn1:.4f}")

y_fit_sn1 = sigmoid(chol_fracs_all, *popt_sn1) # use the optimized parameters

popt_sn2, pcov_sn2 = curve_fit(sigmoid, chol_fracs, lower_upper_means_15pt_sn2_av, p0=initial_guesses)

###sn2


a_opt_sn2, b_opt_sn2, c_opt_sn2 , d_opt_sn2 = popt_sn2
print(f"Optimized parameters: a={a_opt_sn2:.4f}, b={b_opt_sn2:.4f}, c={c_opt_sn2:.4f}")

y_fit_sn2 = sigmoid(chol_fracs_all, *popt_sn2) # use the optimized parameters




```
