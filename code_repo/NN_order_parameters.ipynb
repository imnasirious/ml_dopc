```python
dopc_65_pos_raw = clf_lo_ld_by_chain(sys_name_65 , sys_type , "raw" , "nn" , "sn1" , 3 , start_frame , stop_frame).bilayer_all_dopc_data()
dopc_65_pos_raw_lower , dopc_65_pos_raw_upper = dopc_65_pos_raw

dopc_65_sn1_preds_lower , dopc_65_sn1_pred_upper = dopc_65_11pt_sn1_prediction_labels
dopc_65_sn2_preds_lower , dopc_65_sn2_pred_upper = dopc_65_11pt_sn2_prediction_labels

# shape: (n_frames, n_lipids, 16)
dopc_65_sn1_ops_lower = np.empty([stop_frame, 150, 16])
dopc_65_sn1_ops_upper = np.empty([stop_frame, 150, 16])
dopc_65_sn2_ops_lower = np.empty([stop_frame, 150, 16])
dopc_65_sn2_ops_upper = np.empty([stop_frame, 150, 16])

for i in range(stop_frame):
    for j in range(150):
        dopc_65_sn1_ops_lower[i, j] = DOPC_op_H_av(dopc_65_pos_raw_lower[i, j], "sn1")
        dopc_65_sn1_ops_upper[i, j] = DOPC_op_H_av(dopc_65_pos_raw_upper[i, j], "sn1")
        dopc_65_sn2_ops_lower[i, j] = DOPC_op_H_av(dopc_65_pos_raw_lower[i, j], "sn2")
        dopc_65_sn2_ops_upper[i, j] = DOPC_op_H_av(dopc_65_pos_raw_upper[i, j], "sn2")
#chol_65_pos_raw = clf_lo_ld_by_chain(sys_name_65 , sys_type , "raw" , "nn" , "sn1", 3 , start_frame , stop_frame).bilayer_all_chol_data()

dopc_65_sn1_ops_lower_11pt = convolve_ops(dopc_65_sn1_ops_lower, 11)
dopc_65_sn1_ops_upper_11pt = convolve_ops(dopc_65_sn1_ops_upper, 11)
dopc_65_sn2_ops_lower_11pt = convolve_ops(dopc_65_sn2_ops_lower, 11)
dopc_65_sn2_ops_upper_11pt = convolve_ops(dopc_65_sn2_ops_upper, 11)

lo_ops_sn1_lower_65, ld_ops_sn1_lower_65 = split_ops_by_label(dopc_65_sn1_ops_lower_11pt, dopc_65_sn1_preds_lower)
lo_ops_sn1_upper_65, ld_ops_sn1_upper_65 = split_ops_by_label(dopc_65_sn1_ops_upper_11pt, dopc_65_sn1_pred_upper)
lo_ops_sn2_lower_65, ld_ops_sn2_lower_65 = split_ops_by_label(dopc_65_sn2_ops_lower_11pt, dopc_65_sn2_preds_lower)
lo_ops_sn2_upper_65, ld_ops_sn2_upper_65 = split_ops_by_label(dopc_65_sn2_ops_upper_11pt, dopc_65_sn2_pred_upper)




sn1_lo_mean_65, sn1_lo_std_65, sn1_ld_mean_65, sn1_ld_std_65 = mean_op_profile(
    lo_ops_sn1_upper_65, lo_ops_sn1_lower_65,
    ld_ops_sn1_upper_65, ld_ops_sn1_lower_65
)

sn2_lo_mean_65, sn2_lo_std_65, sn2_ld_mean_65, sn2_ld_std_65 = mean_op_profile(
    lo_ops_sn2_upper_65, lo_ops_sn2_lower_65,
    ld_ops_sn2_upper_65, ld_ops_sn2_lower_65
)
```

```python
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
carbons = np.arange(1, 17)
fig, axes = plt.subplots(3, 2, figsize=(30, 27), sharey=True)
systems = [
    (sn1_lo_mean_90, sn1_ld_mean_90, sn2_lo_mean_90, sn2_ld_mean_90, r'$\mathbf{\chi_C = 0.10}$'),
    (sn1_lo_mean_75, sn1_ld_mean_75, sn2_lo_mean_75, sn2_ld_mean_75, r'$\mathbf{\chi_C = 0.25}$'),
    (sn1_lo_mean_65, sn1_ld_mean_65, sn2_lo_mean_65, sn2_ld_mean_65, r'$\mathbf{\chi_C = 0.35}$'),
]
labels = ['c', 'd', 'e', 'f', 'g', 'h']
label_idx = 0
for row, (sn1_lo, sn1_ld, sn2_lo, sn2_ld, chi) in enumerate(systems):
    axes[row, 0].plot(carbons, sn1_lo, 'o-', color='red', label='Lo', linewidth=4, markersize=10)
    axes[row, 0].plot(carbons, sn1_ld, 'o-', color='green', label='Ld', linewidth=4, markersize=10)
    axes[row, 0].set_ylabel(r'$\mathbf{S_{CD}}$', fontsize=35, fontweight='bold')
    axes[row, 0].set_title(rf'$\mathbf{{sn}}$-1 | {chi}', fontsize=35, fontweight='bold')
    axes[row, 0].set_xticks([2, 4, 6, 8, 10, 12, 14, 16])
    axes[row, 0].tick_params(axis='both', which='major', labelsize=35, width=2.5, length=8, direction="inout")
    for label in axes[row, 0].get_xticklabels() + axes[row, 0].get_yticklabels():
        label.set_fontweight('bold')
    axes[row, 0].text(0.5, 0.98, labels[label_idx], transform=axes[row, 0].transAxes, 
                      fontsize=48, fontweight='bold', ha='center', va='top', color='darkblue')
    legend0 = axes[row, 0].legend(fontsize=25, prop={'weight':'bold', 'size': 30}, frameon=True, fancybox=True)
    for line in legend0.get_lines():
        line.set_linewidth(5.0)
    label_idx += 1
    
    axes[row, 1].plot(carbons, sn2_lo, 'o-', color='red', label='Lo', linewidth=4, markersize=10)
    axes[row, 1].plot(carbons, sn2_ld, 'o-', color='green', label='Ld', linewidth=4, markersize=10)
    axes[row, 1].set_title(rf'$\mathbf{{sn}}$-2 | {chi}', fontsize=35, fontweight='bold')
    axes[row, 1].set_xticks([2, 4, 6, 8, 10, 12, 14, 16])
    axes[row, 1].tick_params(axis='both', which='major', labelsize=35, width=2.5, length=8, direction="inout")
    for label in axes[row, 1].get_xticklabels() + axes[row, 1].get_yticklabels():
        label.set_fontweight('bold')
    axes[row, 1].text(0.5, 0.98, labels[label_idx], transform=axes[row, 1].transAxes, 
                      fontsize=48, fontweight='bold', ha='center', va='top', color='darkblue')
    legend1 = axes[row, 1].legend(fontsize=25, prop={'weight':'bold', 'size': 30}, frameon=True, fancybox=True)
    for line in legend1.get_lines():
        line.set_linewidth(5.0)
    label_idx += 1
# x labels only on bottom row
axes[2, 0].set_xlabel(r'Carbon Index', fontsize=35, fontweight='bold')
axes[2, 1].set_xlabel(r'Carbon Index', fontsize=35, fontweight='bold')
for ax in axes.flat:
    ax.set_ylim(0, 0.30)
plt.tight_layout()
plt.show()
```
